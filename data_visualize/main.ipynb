```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import os
from matplotlib.colors import LinearSegmentedColormap
```

```python
# 从CSV文件读取数据
df = pd.read_csv('../CSV_results/predictability_visualize.csv', index_col=0)

# 创建图表
plt.figure(figsize=(14, 8))

# 设置位置
x_pos = np.arange(len(df.index))

# 绘制点 - 仅用形状区分
plt.scatter(x_pos, df['O-Data & Close'], facecolors='none', edgecolors='black', marker='o', s=50, label='O-Data & Close', zorder=5)  # 空心圆
plt.scatter(x_pos, df['T-Data & Close'], color='black', marker='D', s=50, label='T-Data & Close', zorder=5)  # 实心菱形

# 绘制连接线（从O-Data指向T-Data）
for i, model in enumerate(df.index):
    o_val = df.loc[model, 'O-Data & Close']
    t_val = df.loc[model, 'T-Data & Close']
    
    # 确定线段颜色
    if t_val > o_val:  # 数值上升
        color = 'red'
    else:  # 数值下降
        color = 'green'
    
    # 绘制带箭头的线段（从O-Data指向T-Data）
    plt.annotate('', 
                 xy=(i, t_val),          # 箭头终点（T-Data）
                 xytext=(i, o_val),      # 箭头起点（O-Data）
                 arrowprops=dict(arrowstyle='->', 
                                color=color, 
                                lw=2,
                                shrinkA=5,  # 箭头起点缩进
                                shrinkB=5)  # 箭头终点缩进
                )

# 设置图表属性
plt.xlabel('Models', fontsize=12, fontweight='bold')
plt.ylabel('Predictability (%)', fontsize=12, fontweight='bold')
plt.title('Model Predictability Comparison: O-Data vs T-Data (Close)\n(Arrows show direction from O-Data to T-Data)', 
          fontsize=14, fontweight='bold')
plt.xticks(x_pos, df.index, rotation=45, ha='right')
plt.legend()

# 去掉三边的边框
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['left'].set_visible(False)

# 调整布局
plt.tight_layout()

# 创建image文件夹如果不存在
os.makedirs('../image', exist_ok=True)

# 保存图片到image文件夹
plt.savefig('../image/model_predictability_close.svg', bbox_inches='tight')

# 显示图表
plt.show()
```

```python
# 从CSV文件读取数据
df = pd.read_csv('../CSV_results/predictability_rank_visualize.csv', index_col=0)

# 创建图表
plt.figure(figsize=(14, 8))

# 设置位置
x_pos = np.arange(len(df.index))

# 定义四种数据类别的标记和颜色
markers = ['o', 's', '^', 'D']  # 圆形、方形、三角形、菱形
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728']  # 更鲜明的颜色
categories = df.columns.tolist()


# 为每种数据类别绘制散点图
for i, category in enumerate(categories):
    plt.scatter(x_pos, df[category], 
                marker=markers[i], 
                color=colors[i], 
                s=120, 
                label=category,
                alpha=0.9,
                edgecolors='white',
                linewidth=1.5,
                zorder=5)

# 设置图表属性
min_rank = df.values.min()
max_rank = df.values.max()
y_ticks = np.arange(min_rank, max_rank + 1)
plt.yticks(y_ticks)
plt.xlabel('Models', fontsize=12, fontweight='bold')
plt.ylabel('Rank of Models', fontsize=12, fontweight='bold')
plt.title('Model Predictability Rankings: Comparison Across 4 Datasets', 
          fontsize=14, fontweight='bold')
plt.xticks(x_pos, df.index, rotation=45, ha='right')

# 添加图例
plt.legend()

# 添加网格线
plt.grid(True, alpha=0.2, axis='y')

# 反转y轴，因为排名1是最好的
plt.gca().invert_yaxis()

# 美化边框
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['left'].set_alpha(False)

# 调整布局
plt.tight_layout()

# 创建image文件夹如果不存在
os.makedirs('../image', exist_ok=True)

# 保存图片
plt.savefig('../image/model_predictability_rank.svg', bbox_inches='tight')

# 显示图表
plt.show()
```

```python
# 从CSV文件读取数据
df = pd.read_csv('../CSV_results/predictability_visualize.csv', index_col=0)

# 按AVG_RANK排序，便于观察
df_sorted = df.sort_values('AVG_RANK')

# 创建画布和子图
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20, 8))

# 设置统一的色阶
vmin = df_sorted['AVG_VALUE'].min()
vmax = df_sorted['AVG_VALUE'].max()
norm = plt.Normalize(vmin, vmax)

# 创建从红色（高）到绿色（低）的自定义颜色映射
colors = ['green', 'yellow', 'red']  # 低到高：绿→黄→红
cmap = LinearSegmentedColormap.from_list('red_green', colors, N=100)

# 子图1：柱状图 - AVG_VALUE（使用统一色阶）
bars = ax1.bar(range(len(df_sorted)), df_sorted['AVG_VALUE'], 
               color=cmap(norm(df_sorted['AVG_VALUE'])),
               alpha=0.8, edgecolor='black', linewidth=1)

# 在柱子上添加数值标签
for i, (model, value) in enumerate(zip(df_sorted.index, df_sorted['AVG_VALUE'])):
    ax1.text(i, value + 0.1, f'{value:.2f}', ha='center', va='bottom', 
             fontweight='bold', fontsize=9)

ax1.set_xlabel('Models', fontsize=12, fontweight='bold')
ax1.set_ylabel('Average Predictability', fontsize=12, fontweight='bold')
ax1.set_title('Average Predictability of Each Model\n(Red = High, Green = Low)', 
              fontsize=14, fontweight='bold')
ax1.set_xticks(range(len(df_sorted)))
ax1.set_xticklabels(df_sorted.index, rotation=45, ha='right')
ax1.set_ylim(0, df_sorted['AVG_VALUE'].max() + 1)
ax1.grid(True, alpha=0.3, axis='y')

# 子图2：散点图 + 详细标注 - AVG_RANK（使用统一色阶）
scatter = ax2.scatter(range(len(df_sorted)), df_sorted['AVG_RANK'], 
                     s=200, c=df_sorted['AVG_VALUE'], 
                     cmap=cmap, norm=norm, alpha=0.8, 
                     edgecolors='black', linewidth=1.5)

# 添加详细标注
for i, (model, rank, value) in enumerate(zip(df_sorted.index, df_sorted['AVG_RANK'], df_sorted['AVG_VALUE'])):
    ax2.annotate(f'{model}\nRank: {rank}\nValue: {value:.2f}%', 
                xy=(i, rank),
                xytext=(10, 15),
                textcoords='offset points',
                fontsize=8,
                bbox=dict(boxstyle='round,pad=0.3', facecolor='lightblue', alpha=0.7),
                arrowprops=dict(arrowstyle='->', connectionstyle='arc3,rad=0', color='gray'))

ax2.set_xlabel('Models', fontsize=12, fontweight='bold')
ax2.set_ylabel('Average Rankings', fontsize=12, fontweight='bold')
ax2.set_title('Model Rankings with Predictability Values\n(Red = High, Green = Low)', 
              fontsize=14, fontweight='bold')
ax2.set_xticks(range(len(df_sorted)))
ax2.set_xticklabels(df_sorted.index, rotation=45, ha='right')
ax2.set_ylim(0, df_sorted['AVG_RANK'].max() + 1)
ax2.grid(True, alpha=0.3)
ax2.invert_yaxis()  

# 添加统一的颜色条（放在两个子图右侧）
cbar = plt.colorbar(scatter, ax=ax2)
cbar.set_label('Predictability Value (%)\n(Red = High, Green = Low)', fontsize=10)

plt.tight_layout()

# 创建image文件夹如果不存在
os.makedirs('../image', exist_ok=True)

# 保存图片
plt.savefig('../image/predictability_value_vs_rank.svg', bbox_inches='tight')

plt.show()
```
