```python
from transformer import Transformer, TFLoadData, PositionalEncoding
import pandas as pd
import warnings
from rolling_train_test import RollingTrainTest
warnings.simplefilter(action='ignore', category=FutureWarning)
```

```python
factor = pd.read_csv('../CSV/factor_ma12.csv')
input_dim = factor.shape[1] - 2
print(f"Input dimension: {input_dim}")
```

```python
label = pd.read_csv('../CSV/label_cleaned.csv')
label.describe()
```

```python
Data = TFLoadData(factor, label, batch_size=32, shuffle=True, num_workers=0, train_size=0.5, test_size=0.1)
model_list = [
    Transformer(input_dim, d_model=128, target=3, seq_len=60, nhead=1, num_layers=3, d_ff=256, p=10000,model_name='Transformer')
]
```

```python
count = 1
for model in model_list:
    RTT = RollingTrainTest(model, Data, train_size=0.5, test_size=0.1, epochs=10, patience=3, criterion=None, count=count)
    RTT.info(
        predictability_name = '[Transformer | d_model=128 | seq_len=60 | nHead=1 | num_layers=3 | d_ff=256 | p=10000]'
        )
    RTT.run()
    RTT.backtest(trade_mode=1)
    count += 1
    print(f"Model {model.model_name} backtest completed...")
    print("=" * 50)
```
