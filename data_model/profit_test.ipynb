```python
'''close; clean; mse'''
factor = pd.read_csv('../CSV/factor_cleaned.csv')
input_dim = factor.shape[1] - 2
# print(f"Input dimension: {input_dim}")

label = pd.read_csv('../CSV/label_cleaned.csv')
# label.describe()
target = 3

# load data and model
Data = LoadData(factor, label, batch_size=32, num_workers=0, shuffle=True)
model_list = [
    LinearRegression(input_dim, target=target, model_name="LinearRegression"),
    ElasticNet(input_dim, target=target, alpha=0.5, l1_ratio=1, model_name="Lasso"),
    ElasticNet(input_dim, target=target, alpha=0.8, l1_ratio=0.5, model_name="ElasticNet"),
    RandomForest(n_estimators=100, target=target, max_depth=5, model_name="RandomForest"),
    XGBoost(target=target, n_estimators=100, max_depth=5, learning_rate=0.05, model_name="XGBoost"),
    CNN(input_dim, target=target, alpha=0.8, l1_ratio=0.5, model_name="CNN"),
    NN(input_dim, target=target, alpha=0.8, l1_ratio=0.5, layer=1, model_name="NN"),
    NN(input_dim, target=target, alpha=0.8, l1_ratio=0.5, layer=2, model_name="NN"),
    NN(input_dim, target=target, alpha=0.8, l1_ratio=0.5, layer=3, model_name="NN"),
    NN(input_dim, target=target, alpha=0.8, l1_ratio=0.5, layer=4, model_name="NN"),
    NN(input_dim, target=target, alpha=0.8, l1_ratio=0.5, layer=5, model_name="NN"),
    K_Means_NN(input_dim, target=target, alpha=0.8, l1_ratio=0.5, n_clusters=10, layer=5, model_name="K_Means_NN"),
    Encoder(input_dim, target=target, nlayer=1, flayer=4, d_model=32, nhead=1, dropout=0.1, alpha=0.8, l1_ratio=0.5, model_name="Encoder")
]

# rolling test
count = 0
for model in model_list:
    RTT = RollingTrainTest(model, Data, train_size=0.5, test_size=0.1, epochs=20, patience=3, criterion=None, count=count)
    RTT.info(
        predictability_name = '[--clean--|--close--|--mse--|--epochs=20--]'
        )
    RTT.run()
    RTT.backtest(trade_mode=1)
    count += 1
    print(f"Model {model.model_name} backtest completed...")
    print("=" * 50)
```
